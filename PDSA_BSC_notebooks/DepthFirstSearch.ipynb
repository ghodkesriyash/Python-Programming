#Depth First Search


```python
def neighbours(AMat, i):
  neighbours = []
  (rows, cols) = AMat.shape

  for j in range(cols):
    if AMat[i, j] == 1:
      neighbours.append(j)

  return neighbours
```

```python
def DFSinit(Amat):
    (rows,cols) = Amat.shape
    visited = {}
    parent = {}

    for i in range(cols):
        visited[i] = False
        parent[i] = -1

    return visited,parent


def DFS(Amat,v,visited,parent):
    visited[v] = True

    for k in neighbours(Amat, v):
        if not visited[k]:
            parent[k] = v
            (visited, parent) = DFS(Amat, visited, parent, k)

    return (visited, parent)
```
