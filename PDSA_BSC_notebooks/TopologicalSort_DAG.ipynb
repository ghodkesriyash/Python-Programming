#Topological Sort on DAG
- Time  complexity ==> O(n^2)

```python
def topo_sort(AMat):
  (rows, cols) = AMat.shape
  indegree = {}
  topo_sort_list = []

  for c in range(cols):
    indegree[c] = 0

    for r in range(rows):
      if AMat[r, c] == 1:
        indegree[c] += 1

  for i in range(rows):
    j = min([k for k in range(cols)
            if indegree[k] == 0])
    topo_sort_list.append(j)
    indegree[j] -= 1

    for k in range(cols):
      if AMat[j, k] == 1:
        indegree[k] -= 1

  return topo_sort_list
```
