# Searching
## 1. Linear Search ( naive)
 - Complexity ==> O(n)
  

```python
data = [47, 83, 12, 65, 91, 29, 56, 74, 38, 10, 99, 22, 61, 85, 43, 17, 70, 34, 58, 96]
```

```python
def LinearSearch(arr : list,elem : int):
  for x in range(len(arr)):
    if arr[x] == elem:
      return True
  return False
```

```python
LinearSearch(data,38)
```

## 2.Binary Search
- Complexity ==> O(logn)
- List needs to be sorted beforehand

```python
def BinarySearch(arr : list, elem : int):
    mid = len(arr) // 2
    if arr == []:
        return False
    if arr[mid] == elem:
        return True
    elif elem < arr[mid]:
        return BinarySearch(arr[:mid],elem)
    else:
        return BinarySearch(arr[mid+1:],elem)
```

```python
data.sort()
BinarySearch(data,91)
```

#Sorting
## 1. Selection Sort
 - complexity ==> o(n^2)

```python
def SelectionSort(arr : list):
    n = len(arr)
    if n < 1:
        return (arr)
    for i in range(n):
        min_idx = i
        for j in range(i +1,n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i],arr[min_idx] = arr[min_idx],arr[i]
    return arr
```

```python
SelectionSort(data)
```

##2. Insertion Sort
- Complecity ==> O(n^2)

```python
def InsertionSort(arr:list):
    n = len(arr)
    if n < 1:
        return arr
    for i in range(n):
        j = i
        while ( j > 0 and arr[j] < arr[ j -1 ]):
            arr[j],arr[j-1] = arr[j-1],arr[j]
            j -= 1
    return arr
```

```python
InsertionSort(data)
```

## 3.Merge Sort
- Based on divide and conquor principal
- complexity ==> O(n logn)

```python
def merge(L:list,R:list):
    (m,n) = (len(L),len(R))
    (merged,i,j,k) = ([],0,0,0)
    while k < m + n:
        if i == m:
            merged.extend(R[j:])
            k = k + (n - j)
        elif j == n:
            merged.extend(L[i:])
            k = k + (m - i)
        elif(L[i] < R[j]):
            merged.append(L[i])
            i += 1
            k += 1
        else:
            merged[k].append(R[j])
            j += 1
            k += 1
    return merged

def MergeSort(arr:list):
    n = len(arr)
    if n <= 1:
        return arr
    mid = n // 2
    L = MergeSort(arr[:mid])
    R = MergeSort(arr[mid + 1:])
    B = merge(L,R)

    return B
```

```python
MergeSort(data)
```

##Quick Sort
- compplexity ==> O(n^2) [worst case] average is O(nlogn)
- Also follows divide and conquor

```python
def QuickSort(arr:list,l:int,r:int): #We are sorting the slice of array arr[l:r]
    if (r - l <= 1):
        return arr
    (pivot,lower,upper) = (arr[l],l+1,l+1)
    for i in range(l+1,r):
        if arr[i] > pivot:
            upper += 1
        else:
            arr[i],arr[lower] = arr[lower],arr[i]
            upper += 1
            lower += 1
    arr[l],arr[lower -1] = arr[lower - 1],arr[l] #move pivot between upper and lower
    lower -= 1
    QuickSort(arr,l,lower)
    QuickSort(arr,lower+1,upper)
```

```python
QuickSort(data,0,-1)
```
